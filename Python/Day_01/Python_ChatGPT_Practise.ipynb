```python
employees = [
    {"name":"Ravi","salary":80000,"department":"IT"},
    {"name":"Amit","salary":45000,"department":"HR"},
    {"name":"Neha","salary":120000,"department":"Finance"},
    {"name":"Karan","salary":70000,"department":"IT"}
]
```

```python
#Print only employees earning more than ₹70,000.

print("Name of employees earning more than ₹70,000: ")
for i in employees:
    if i["salary"]>70000:
        print(i["name"])
    
```
